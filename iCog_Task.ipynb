# **Create a Quantum Circuit with a Hadamard Gate**

- Write a Qiskit program to create a single-qubit circuit.
- Apply a Hadamard gate to the qubit.
- Measure the qubit and visualize the result using a histogram.

```python

# Import necessary libraries from the Qiskit ecosystem.
import qiskit
from qiskit.visualization import plot_histogram
from qiskit_aer.primitives import SamplerV2



# Create a quantum circuit named 'qc' with a single qubit.
qc = qiskit.QuantumCircuit(1)



# Apply a Hadamard gate to qubit 0, placing it in a superposition state.
# The Hadamard gate creates an equal probability of measuring 0 or 1.
qc.h(0)



# Add a measurement operation to the circuit to measure all qubits.
# This is necessary to obtain results from the quantum computation.
qc.measure_all()



# Visualize the constructed quantum circuit using a textual representation.
# This helps in understanding the sequence of operations applied to the qubit.
qc.draw()


```

```python


# Setting up a quantum simulator using SamplerV2 from Qiskit Aer.
# This simulator mimics an ideal, noise-free quantum computer.
sampler = SamplerV2()




# Executing the quantum circuit 'qc' on the simulator.
# 'shots=128' indicates that the circuit will be run 128 times
# to gather statistical results.
job = sampler.run([qc], shots=128)



# Retrieving the results of the simulation.
result = job.result()



# Extracting the measurement outcomes and their counts.
# 'counts' will be a dictionary where keys represent measurement outcomes
# (e.g., '0' or '1') and values represent the number of times each outcome was observed.
counts = result[0].data.meas.get_counts()




# Displaying the measurement results in the console.
print('counts(ideal):', counts)




# Generating a histogram to visualize the measurement results.
# The histogram provides a clear view of the probability distribution
# of the quantum circuit's output.
plot_histogram(counts)



```

# **Implement a Quantum Coin Flip**
- Simulate a fair coin flip using a quantum circuit.
- Print the probability of getting Heads or Tails.

```python
#insert the code
# Create a quantum circuit with 1 qubit
qc = qiskit.QuantumCircuit(1)



# Apply a Hadamard gate to put the qubit in superposition
qc.h(0)




# Measure the qubit
qc.measure_all()



# Use the Aer simulator to run the circuit
sampler = SamplerV2()


job = sampler.run([qc], shots=1024)  # Increase shots for better statistics
result = job.result()
counts = result[0].data.meas.get_counts()



# Print the probabilities
print("Probability of Heads (0):", counts.get('0', 0) / 1024)
print("Probability of Tails (1):", counts.get('1', 0) / 1024)
```

# **Implement a Bell State (Entanglement)**
- Create a two-qubit circuit.
- Apply a Hadamard gate to the first qubit and a CNOT gate between the two qubits.
- Measure both qubits and verify if they are always correlated.

```python
#insert your code here
# Create a quantum circuit with 2 qubits
qc = qiskit.QuantumCircuit(2)

# Apply a Hadamard gate to the first qubit
qc.h(0)

# Apply a CNOT gate with the first qubit as control and the second as target
qc.cx(0, 1)

# Measure both qubits
qc.measure_all()

# Use the Aer simulator to run the circuit
sampler = SamplerV2()
job = sampler.run([qc], shots=1024)  # Increase shots for better statistics
result = job.result()
counts = result[0].data.meas.get_counts()

# Print the results
print(counts)

# Visualize the circuit
qc.draw()
```

### **How does superposition affect the information processing capabilities of a quantum computer?**


Superposition gives quantum computers a huge advantage by allowing their basic units of information, called qubits, to exist in multiple states at once. Unlike classical computers, where bits can only be either 0 or 1 at a time, qubits can be both 0 and 1 simultaneously. This unique property boosts a quantum computer’s processing power in several ways:

###How do quantum gates manipulate interference to solve problems faster than classical computers?



Quantum gates make quantum computers faster by using interference to focus on the right answers and get rid of the wrong ones. Here's how it works:

* **Superposition:** A quantum gate can make a
qubit be in both 0 and 1 states at the same time, allowing the computer to check multiple possibilities at once.
* **Interference:**  Quantum gates control how the different states interact, boosting the chances of correct answers and canceling out incorrect ones.
* **Speeding up the solution:** By adjusting the interference, the computer can find the right answer faster than a regular computer, which checks one possibility at a time.
This is how quantum computers can solve certain problems much quicker than classical computers.

### How does measurement collapse a quantum state, and why is it important in quantum computing algorithms?

In quantum mechanics, measurement causes a quantum state to "collapse." Before measurement, a system can be in multiple states at once (a superposition), like a coin spinning in the air, being both heads and tails. When you measure it, the system "chooses" one state.

**How It Works**

* **Superposition:** A quantum system exists in a combination of states, such as being both 0 and 1 at the same time.
* **Measurement:** When you measure the system, it collapses to one of its possible states. The chances of it collapsing to a particular state depend on how the system was set up.
* **Probabilistic:** The result is not guaranteed but based on probability. For instance, a 50% chance of being 0 and 50% chance of being 1.

**Why It Matters in Quantum Computing?**
* **Parallelism: **Before measurement, quantum computers can explore multiple possibilities at once. This makes quantum algorithms, like Shor's for factoring and Grover's for searching, much faster than classical ones.

* **Uncertain Results:** Since measurements are probabilistic, we often need to make multiple measurements to increase the chances of getting the correct answer.

* **Interference:** Quantum algorithms use superposition to increase the chances of measuring a correct result. They interfere with wrong answers to cancel them out.

* **Entanglement:** Measurement is also key in entanglement, where measuring one qubit instantly affects another, no matter how far apart they are. This is useful for things like secure communication.
